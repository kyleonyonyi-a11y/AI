# Healthcare Stroke Dataset Analysis
This Google Colab notebook loads and explores the healthcare stroke dataset. Upload `healthcare-dataset-stroke-data-selected-columns.csv` when prompted.

```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd
import io

# Load the uploaded CSV file
file_name = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print('Dataset loaded successfully!')
print('Rows and columns:', df.shape)
display(df.head())
```

## Dataset overview

```python
display(df.info())
display(df.describe(include='all'))
print('\nMissing values per column:')
display(df.isnull().sum())
```

## Data preparation

```python
# Remove duplicate rows, if any
df = df.drop_duplicates()

# Display the updated dataset size and missing values
print('Updated shape:', df.shape)
display(df.isnull().sum())
```

## Explore the stroke outcome

```python
import matplotlib.pyplot as plt

if 'stroke' in df.columns:
    print(df['stroke'].value_counts())
    df['stroke'].value_counts().sort_index().plot(kind='bar')
    plt.title('Distribution of Stroke Outcome')
    plt.xlabel('Stroke (0 = No, 1 = Yes)')
    plt.ylabel('Number of records')
    plt.xticks(rotation=0)
    plt.show()
else:
    print("A 'stroke' column was not found. Available columns:", list(df.columns))
```
